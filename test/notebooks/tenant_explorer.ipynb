```python
import sys
from pathlib import Path

# Locate project directory
PROJECT_DIR = Path.cwd()

if not (PROJECT_DIR / "tenant_explorer.py").exists():
    PROJECT_DIR = PROJECT_DIR.parent

sys.path.insert(0, str(PROJECT_DIR))

from tenant_explorer import TenantExplorer

import pandas as pd
from IPython.display import display, Markdown
```

```python
explorer = TenantExplorer()

print("Tenant Explorer initialized")
print("Tenant ID:", explorer.tenant_id)
print("Database:", explorer.db_path)
```

```python
sync_results = explorer.sync()

display(sync_results)
```

```python
display(Markdown("# Microsoft 365 Tenant Dashboard"))

display(explorer.dashboard().style
    .hide(axis="index")
    .format({"Value": "{:,}"}))
```

```python
organization = explorer.organization()

display(organization[[
    "tenant_id",
    "display_name",
    "country",
    "created_at"
]])
```

```python
users = explorer.users()

display(
    users.style
    .hide(axis="index")
    .format({
        "account_enabled": lambda x:
            "Enabled" if x == 1 else
            "Disabled" if x == 0 else "Unknown"
    })
)
```

```python
groups = explorer.groups()

display(groups)
```
